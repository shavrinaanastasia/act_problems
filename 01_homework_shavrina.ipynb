## Задание 5 (1)

Напишите такую регулярку, которая извлечет ФИО в каноническом виде из примеров ниже.   
"Иванов И. И." -> должно нормализоваться до "Иванов И.И." (Фамилия + Инициалы без пробелов)  
"И. И. Иванов" - "Иванов И.И."

Подсказка: напишите отдельные регулярки для инициалов и фамилий, затем напишите еще одну регулярку, которая включает возможные комбинации и порядки. Используйте именованные группы, чтобы преобразовать части в каноничный порядок.
В коде не должно быть никаких дополнительных нормализаций (дополнительный re.sub или .replace и т.п)

```python
# Инициалы и фамилию задаём отдельно, а затем собираем два допустимых
# порядка: «Фамилия И. О.» и «И. О. Фамилия». Вторая часть фамилии нужна
# для примера «Ван Дамм»; второй инициал может отсутствовать.
INITIAL = r"[А-ЯЁ]\."
SURNAME = r"[А-ЯЁ][а-яё]+(?:\s+[А-ЯЁ][а-яё]+)*"

name_re = re.compile(rf"""
    ^
    (?:
        (?P<surname_first>{SURNAME})
        \s+
        (?P<after_i1>[А-ЯЁ])\.\s*
        (?:(?P<after_i2>[А-ЯЁ])\.)?
        |
        (?P<before_i1>[А-ЯЁ])\.\s*
        (?:(?P<before_i2>[А-ЯЁ])\.\s*)?
        (?P<surname_last>{SURNAME})
    )
    $
""", re.VERBOSE)


def canonical_name(match):
    """Собирает каноническую запись только из групп одного выражения."""
    surname = match.group("surname_first") or match.group("surname_last")
    initial_1 = match.group("after_i1") or match.group("before_i1")
    initial_2 = match.group("after_i2") or match.group("before_i2")
    second_initial = f"{initial_2}." if initial_2 else ""
    return f"{surname} {initial_1}.{second_initial}"
```

```python
NAMES = [
    "Иванов И. И.",
    "И. И. Иванов",
    "Иванов И.И.",
    "А.С. Пушкин",
    "Пушкин А.С.",
    "Д. М. Шеберле",
    "М. Кюри",
    "Ж.К. Ван Дамм",
]
```

```python
canonical_names = [name_re.sub(canonical_name, name) for name in NAMES]
canonical_names
```

```python
TEXT = """
В русской типографике существует правило использования кавычек в зависимости от контекста:
Кавычки «ёлочки» используются для основного выделения текста в предложении: прямой речи, терминов, условных обозначений, названий произведений. Это стандартный тип кавычек для русских текстов.
Кавычки „лапки“ обычно используются внутри кавычек-ёлочек, когда необходимо выделить текст внутри уже выделенного текста. Это вложенные кавычки.
Пример использования:
• Основной текст в кавычках: Рекомендую к прочтению роман Льва Толстого — «Война и мир»
• Вложенные кавычки: «Она сказала: „Я встречусь с вами завтра“»

Во французском языке также активно используются «ёлочки» для выделения прямой речи. Собственно, этот тип кавычек называется французским или «chevrons pointés».
В англоязычной типографике двойные кавычки-лапки или «quotation mark» (“ ”) являются стандартом для обозначения прямой речи или цитат. И это правило иногда переносится на латинские тексты (в которых принято использовать «ёлочки»), особенно если оформление латинских текстов схоже с англоязычными текстами. При этом английские кавычки визуально отличаются от русских „кавычек-лапок“.
Запомнить, как должны выглядеть «лапки», довольно просто: в русской версии открывающиеся „лапки“ (пример использования) выглядят как две запятых. В английских кавычках — они всегда сверху, а первая пара похожа на две “шестёрки” (пример использования).
Кавычки могут применяться для оформления названий, например, если они имеют неоднозначное значение или необходимо указать на смысловую связь между словами.
Например: “Yellow Submarine” (песня The Beatles), “The Godfather” (художественный фильм), “The Times They Are a-Changin” (песня Боба Дилана)
Одиночные или марровские кавычки (‘ ’) используются для цитат внутри цитат или для выделения специфических терминов.
"""
```

```python
# Символ “ двусмысленен: по-русски он закрывает кавычки, а по-английски
# открывает их. Для английской открывающей кавычки проверяем, что слева нет
# буквы или цифры, а справа сразу начинается текст. Остальные типы однозначны.
quote_re = re.compile(r"""
    (?P<opening>
        [«„‘]
        |
        (?<![0-9A-Za-zА-Яа-яЁё])“
    )
    |
    (?P<closing>
        [»”’]
        |
        (?<=\S)“
    )
""", re.VERBOSE)

text_with_guillemets = quote_re.sub(
    lambda match: "«" if match.lastgroup == "opening" else "»",
    TEXT,
)
text_with_guillemets
```
